Import

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import make_column_transformer
```

Data Prep

```python
essays = pd.read_csv('kdd-cup-2014-predicting-excitement-at-donors-choose/data/essays.csv')
outcomes = pd.read_csv('kdd-cup-2014-predicting-excitement-at-donors-choose/data/outcomes.csv')
projects = pd.read_csv('kdd-cup-2014-predicting-excitement-at-donors-choose/data/projects.csv')
resources = pd.read_csv('kdd-cup-2014-predicting-excitement-at-donors-choose/data/resources.csv')
```

```python
#get length of project description

essays['short_description'] = essays['short_description'].astype(str)
essays['essay'] = essays['essay'].astype(str)
essays['short_description_length'] = essays['short_description'].apply(len)
essays['essay_length'] = essays['essay'].apply(len)

#make project outcome binary

to_binary = {'f':0,'t':1}
outcomes['fully_funded'] = outcomes['fully_funded'].map(to_binary)
```

```python
project_outcomes = projects.merge(outcomes,how='inner',on='projectid')
essay_outcomes = essays.merge(outcomes,how='inner',on='projectid')
```

Feature Matrix 1

```python
df = project_outcomes[['projectid','school_state',"school_metro","fully_funded",'poverty_level']]
```

```python
df = df[df['school_state'] != 'La']
df.dropna()
```

```python
ohe = OneHotEncoder(sparse_output=False)
ct = make_column_transformer(
       (ohe, ['school_state','school_metro','poverty_level']),
       remainder = 'passthrough'
)
```

```python
matrix = ct.fit_transform(df)
matrix[:5, : ]
```

Feature Matrix 2

```python
df2 = project_outcomes[['projectid','total_price_excluding_optional_support','students_reached','fully_funded']]
```

```python
q1 = df2['total_price_excluding_optional_support'].quantile(0.25)
q3 = df2['total_price_excluding_optional_support'].quantile(0.75)
iqr = q1 - q3

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
```

```python
df2_iqr = df2[(df2['total_price_excluding_optional_support'] >= lower_bound)]
df2_iqr = df2[(df2['total_price_excluding_optional_support'] <= upper_bound)]
```

```python
df2_iqr.groupby('fully_funded').describe()

#question: which iqr do we take if we have two features with outliers?
```
