```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt

from dataprocessing import data, data_noutliers
```

```python
#with outlier
data[['total_price_excluding_optional_support','students_reached','fully_funded']].groupby('fully_funded').describe()
```

```python
#data without outlier
data_noutliers[['total_price_excluding_optional_support','students_reached','fully_funded']].groupby('fully_funded').describe()
```

```python
#Number Students Reached
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# First 
sns.histplot(data['students_reached'],bins=30, kde=False,ax=axes[0])
axes[0].set_title('Number Students Reached')

# Second 
sns.histplot(data_noutliers['students_reached_scaled'],bins=30, kde=False,ax=axes[1])
axes[1].set_title('Number Students Reached Scaled and Without Outliers')

plt.tight_layout()
```

```python
#Number Students Reached
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# First 
sns.histplot(data['total_price_excluding_optional_support'],bins=30, kde=False,ax=axes[0])
axes[0].set_title('Funding Amount Requested')

# Second 
sns.histplot(data_noutliers['total_price_excluding_optional_support_scaled'],bins=30, kde=False,ax=axes[1])
axes[1].set_title('Funding Amount Requested Scaled and Without Outliers')

plt.tight_layout()
```
